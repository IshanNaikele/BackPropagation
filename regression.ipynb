```python
def initialize_parameters(layers_dims):
    np.random.seed(3) # Generates the Same Random number again & again
    parameters={}
    L=len(layers_dims) # Length of the Neural Network .Ex [2,2,1]-->3

    for l in range(1,L): # Runs from 1 to 2 as L=3
        # Size of Weights is like current layer * previous layer
        # Ex [2,2,1] ---> 2 * 2 =4 wts for 1st layer ,2 * 1 =2 wts for 2nd layer --->All are initialized with 0.1 
        parameters['W'+str(l)]= np.ones((layers_dims[l-1],layers_dims[l]))*0.1   
        # Size of the bias vector is like the number of nodes in current layer 
        # Ex [2,2,1] ---> 2 for 2nd layer and 1 for 3rd layer --->All are initialized with 0
        parameters['b'+str(l)]= np.zeros((layers_dims[l],1))
                                         
    return parameters
```

```python
initialize_parameters([2,2,1])
```

```python
# Deals with Weighted Sum Calculation .
def linear_forward(A_prev,W,b):
    Z=np.dot(W.T,A_prev)+b
    return Z

# Forward Propagation 
# Formula = X * W(Transpose)  + B
```

```python
# Deals with Forward Propagation --->Weighted Sum Calculation ,Passing through Activation Function ,Loss Calculation 
def L_layer_forward(X,parameters):
    # Here ,parameters consists of weights & bias information .
    # The X is the input of the current neuron or we can say that the output of the previous node . 
    A=X
    L=len(parameters)//2 # No of Layers in neural Network .4//2 ==2

    for l in range(1,L+1):# Runs for 1 to 2 as there are two layers in [2,2,1]
        A_prev=A

        W1=parameters['W'+str(l)]
        b1=parameters['b'+str(l)]
        print("A"+str(l-1)+": ",A_prev)
        print("W"+str(l)+": ",W1)
        print("b"+str(l)+": ",b1)
        print("__"*20)

        A=linear_forward(A_prev,W1,b1)
        print("A"+str(l)+": ",A)
        print("***"*20)
    return A,A_prev
```

```python
X = df[['CGPA', 'Profile_Score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
y = df[['LPA']].values[0][0]

# Parameter initialization
parameters = initialize_parameters([2,2,1])

y_hat,A1 = L_layer_forward(X, parameters)
```
